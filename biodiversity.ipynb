```python
df_species["conservation_status"].unique()
```

```python
df_species["conservation_status"].isna().sum()
```

```python
status_counts = df_species["conservation_status"].value_counts()
print(status_counts)
```

```python
print(type(status_counts))
```

```python
#importing matplot library
import matplotlib.pyplot as plt

#plotting a bar graph to show conservation_status distribution
status_counts.plot(kind="bar")
plt.title("Distribution of Conservation Status of Species")
plt.xlabel("Conservation Status")
plt.ylabel("No. of Species")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
